```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
train_df = pd.read_csv('data/train.csv')
```

```python
test_df = pd.read_csv('data/test.csv')
sample_sub = pd.read_csv('data/sample_submission.csv')
```

```python
# Select Features & Target
feature_cols = [
    "season", 'year', 'Month', 'is_holiday', 'weekdaynum', 'is_workingday',
    'weathersituation', 'atemp_normalized', 'humidity_normalized', 'windspeed_normalized'
]

X = train_df[feature_cols]
y = train_df['total_rental']
X_test = test_df[feature_cols]
```

```python
# Model 1: Decision Tree
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeRegressor

dt = DecisionTreeRegressor(max_depth=5, random_state=42)
dt_scores = -cross_val_score(dt, X, y, scoring='neg_root_mean_squared_error', cv=5)

print(f"Decision Tree CV RMSE: {dt_scores.mean():.2f}")
```
